## Rebuilding the Environment with `uv`

This repository's Python environment is managed with [`uv`](https://docs.astral.sh/uv/), a fast, all-in-one Python package and project manager. It reads `pyproject.toml` (the declared dependencies) and `uv.lock` (the exact, pinned versions) to recreate the *same* environment on any machine.

You do **not** need to `pip install` anything manually, and you do **not** need to create a virtual environment yourself: `uv` does both for you.

### 1. Install `uv`

```bash
# macOS / Linux
curl -LsSf https://astral.sh/uv/install.sh | sh

# Windows (PowerShell /!\ No Admin Right /!\)
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

### 2. Clone the repository and sync the environment

```bash
git clone <repository-url>
cd <repository>

# Creates a local .venv/ and installs every dependency pinned in uv.lock
uv sync
```

`uv sync` reads `requires-python` from `pyproject.toml` (Python `>= 3.10`), downloads a matching interpreter if needed, creates `.venv/`, and installs `numpy`, `matplotlib`, `ipykernel` and `jupyterlab` at the exact versions recorded in `uv.lock`. Everyone in class ends up with the identical set of package versions.

### 3. Launch Jupyter

```bash
uv run jupyter lab
```

or, if you prefer VS Code or another editor, select the interpreter inside `.venv/` as the notebook kernel.

### 4. Adding a new dependency later

Don't `pip install` by hand, that would desynchronize your environment from `uv.lock`. Instead:

```bash
uv add <package-name>
```

## Check 3: read a spike file

This is the kind of data you will use in the projects: a CSV file with **one row per spike**, telling in which trial, which neuron fired, and when. Open `projects/Project 1 - BCI/data/bci_visual_2stim_spikes.csv` in a spreadsheet or in VS Code to see it for yourself.

````python
table = np.loadtxt(root / "projects/Project 1 - BCI/data/bci_visual_2stim_spikes.csv", delimiter=",", skiprows=1, dtype=int)
print("columns: trial, neuron, time_ms")
print(table[:5])
print("number of spikes:", len(table))
print("Data OK ✔")
````

## Check 4: plotting

If you see one raster below (one row per trial, one tick per spike), matplotlib works.

````python
import matplotlib.pyplot as plt

neuron = 0
for k in range(20):
    times = table[(table[:, 0] == k) & (table[:, 1] == neuron), 2]
    plt.vlines(times, k, k + 0.8, lw=0.8)
plt.xlabel("time (ms)")
plt.ylabel("trial")
plt.title(f"neuron {neuron}, first 20 trials")
plt.show()
print("Plotting OK ✔")
````
